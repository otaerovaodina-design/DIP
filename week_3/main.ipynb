```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread("nature.jpg")
print(image.shape)

trans_matrix = np.float32([[1,0,500],[0,1,250]])

translated_image = cv2.warpAffine(image, trans_matrix, (image.shape[1], image.shape[0]))
print(translated_image.shape)
plt.imshow(cv2.cvtColor(translated_image, cv2.COLOR_BGR2RGB))
plt.title("translated image")
plt.axis("off")
```

```python
(h, w) = image.shape[:2]
center = (w // 2, h // 2)

rotation_matrix = cv2.getRotationMatrix2D(center, 90, 1.0)

rotated_image = cv2.warpAffine(image, rotation_matrix, (w, h))

print(translated_image.shape)
plt.imshow(cv2.cvtColor(rotated_image, cv2.COLOR_BGR2RGB))
plt.title("rotated image ")
plt.axis("off")
```

```python
Factor = 1.5

scaling_matrix = cv2.getRotationMatrix2D((0,0), 0, Factor)

scaling_image = cv2.warpAffine(image, scaling_matrix, (w, h))

print(translated_image.shape)
plt.imshow(cv2.cvtColor(scaling_image, cv2.COLOR_BGR2RGB))
plt.title("scaled image ")
plt.axis("off")
```

```python
m = 0.33
shearing_matrix = np.float32([[1,m,0],[0,1,0]])

sheared_image = cv2.warpAffine(image, shearing_matrix, (w, h))

plt.imshow(cv2.cvtColor(sheared_image, cv2.COLOR_BGR2RGB))
plt.title('Sheared Image (x-axis, factor)'+str(m)+')')
plt.axis('off')
plt.show()
```
