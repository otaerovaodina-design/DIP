```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Load image
image = cv2.imread("111.jpg")

# Check if image was loaded successfully
if image is None:
    print("Error: Could not load image '111.jpg'. Please ensure the file exists in the correct directory.")
    exit()

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Step 1: Blur to remove noise
blur = cv2.GaussianBlur(gray, (5, 5), 0)

# Step 2: Edge detection
edges = cv2.Canny(blur, 50, 150)

# Step 3: Threshold to highlight crack
_, thresh = cv2.threshold(gray, 100, 255, cv2.THRESH_BINARY_INV)

# Step 4: Morphological operations to enhance crack
kernel = np.ones((3,3), np.uint8)
morph = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel, iterations=2)

# Step 5: Combine edges + threshold
combined = cv2.bitwise_or(edges, morph)

# Step 6: Find contours (damaged areas)
contours, _ = cv2.findContours(combined, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# Draw detected crack
output = image.copy()
for cnt in contours:
    if cv2.contourArea(cnt) > 100:  # filter small noise
        x, y, w, h = cv2.boundingRect(cnt)
        cv2.rectangle(output, (x, y), (x+w, y+h), (0, 0, 255), 2)

# Show results
cv2_imshow(image)
cv2_imshow(edges)
cv2_imshow(output)
```
