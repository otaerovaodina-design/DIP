```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Helper to display images nicely
def show_image(title, img, cmap='gray'):
    plt.imshow(img, cmap=cmap)
    plt.title(title)
    plt.axis('off')
    plt.show()
```

```python
# Load image (use your own path)
img = cv2.imread('eye.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

show_image("Original", cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
show_image("Grayscale", gray)
```

```python
grad_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
grad_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

gradient = np.sqrt(grad_x**2 + grad_y**2)
gradient = cv2.normalize(gradient, None, 0, 255, cv2.NORM_MINMAX)
gradient = gradient.astype(np.uint8)

show_image("Simple Gradient", gradient)
```

```python
sobelx = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sobely = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

sobel = cv2.magnitude(sobelx, sobely)
sobel = cv2.normalize(sobel, None, 0, 255, cv2.NORM_MINMAX)
sobel = sobel.astype(np.uint8)

show_image("Sobel", sobel)
```

```python
laplacian = cv2.Laplacian(gray, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)

show_image("Laplacian", laplacian)
```

```python
canny = cv2.Canny(gray, threshold1=100, threshold2=150)

show_image("Canny Edge Detection", canny)
```

```python
titles = ["Gradient", "Sobel", "Laplacian", "Canny"]
images = [gradient, sobel, laplacian, canny]

plt.figure(figsize=(12, 6))

for i in range(4):
    plt.subplot(2, 2, i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
blur = cv2.GaussianBlur(gray, (5,5), 0)
show_image("Gaussian Blur", blur)
```

```python
laplacian_blur = cv2.Laplacian(blur, cv2.CV_64F)
laplacian_blur = cv2.convertScaleAbs(laplacian_blur)

show_image("Laplacian (Blurred)", laplacian_blur)
```

```python
sobelx_blur = cv2.Sobel(blur, cv2.CV_64F, 1, 0)
sobely_blur = cv2.Sobel(blur, cv2.CV_64F, 0, 1)

sobel_blur = cv2.magnitude(sobelx_blur, sobely_blur)
sobel_blur = cv2.normalize(sobel_blur, None, 0, 255, cv2.NORM_MINMAX)
sobel_blur = sobel_blur.astype(np.uint8)

show_image("Sobel (Blurred)", sobel_blur)
```

```python
canny_blur = cv2.Canny(blur, 100, 200)

titles = ["Gradient", "Sobel", "Laplacian", "Canny"]
images = [gradient, sobel_blur, laplacian_blur, canny_blur]

plt.figure(figsize=(12,6))

for i in range(4):
    plt.subplot(2,2,i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# EYE BLOOD VESSELS DETECTION


img = cv2.imread('eye.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

show_image("Original", img_rgb)

green = img[:, :, 1]
show_image("Green Channel", green)
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
enhanced = clahe.apply(green)

show_image("Enhanced (CLAHE)", enhanced)
blur = cv2.GaussianBlur(enhanced, (5,5), 0)
show_image("Blurred", blur)
```

```python
 sobelx = cv2.Sobel(blur, cv2.CV_64F, 1, 0)
sobely = cv2.Sobel(blur, cv2.CV_64F, 0, 1)

sobel = cv2.magnitude(sobelx, sobely)
sobel = cv2.normalize(sobel, None, 0, 255, cv2.NORM_MINMAX)
sobel = sobel.astype(np.uint8)

show_image("Sobel Vessels", sobel)
```
